2️⃣ Understanding Figure & Axes
Matplotlib works with Figure and Axes:

Figure: The entire plotting area.

Axes: The individual subplots inside the figure.

```python
#✅ Creating a Figure & Adding Axes
fig, ax = plt.subplots()  # Creates a figure and a set of axes
ax.plot(x, y, marker='o')  # Line plot with markers
ax.set_title("Basic Line Plot")  # Adding title
ax.set_xlabel("X-axis Label")  # X-axis label
ax.set_ylabel("Y-axis Label")  # Y-axis label
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot(x, y, marker='o', color='b')
axes[0].set_title("Line Plot")

axes[1].scatter(x, y, color='r')  # Scatter plot
axes[1].set_title("Scatter Plot")

plt.tight_layout()  # Adjust layout
plt.show()
```

```python
# Create a figure and multiple axes using subplots
fig, axes = plt.subplots(2, 2, figsize=(8, 6))  # 2x2 grid of plots

# Plot something on each subplot
axes[0, 0].plot([1, 2, 3], [4, 5, 6])  # Top-left
axes[0, 1].scatter([1, 2, 3], [6, 5, 4])  # Top-right
axes[1, 0].bar([1, 2, 3], [6, 5, 4])  # Bottom-left
axes[1, 1].hist([1, 2, 2, 3, 3, 3, 4, 4, 4, 4])  # Bottom-right

plt.tight_layout()  # Adjusts spacing between subplots
plt.show()
```

```python
Think of matplotlib.pyplot as your artistic toolkit. It allows you to create a Figure (like a blank canvas).

Inside this Figure, you can have one or more Axes (the area where actual data is plotted).

The Axes can be thought of as different sections or frames within the canvas, where different elements (plots) are drawn.

By default, a figure has one set of axes, but you can create multiple ones using subplots, enabling you to render multiple plots in a single figure.
```
